```python
import pandas as pd

expenses = pd.read_csv("datasets/expenses.csv")

expenses.head()
# print(expenses['Date'])
# print(expenses['Date'].dtype)
```

```python
category_summary = expenses.groupby('Category')['Amount'].agg(['sum', 'mean']).reset_index()

category_summary.head()
# expenses.head()
```

```python
seasonal_summary = expenses.groupby('Season')['Amount'].agg(['sum', 'mean']).reset_index()

print(seasonal_summary)

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.bar(seasonal_summary['Season'], seasonal_summary['sum'])
plt.title('Total Expenditure by Season')
plt.xlabel('Season')
plt.ylabel('Total Expenditure (Amount)')
plt.show()

expenses.head()
```

```python
fixed_cost_categories = ['Salaries', 'Rent', 'Utilities']

expenses['Cost_Type'] = expenses['Category'].apply(lambda x: 'Fixed' if x in fixed_cost_categories else 'Variable')

cost_type_summary = expenses.groupby('Cost_Type')['Amount'].agg(['sum', 'mean']).reset_index()

cost_type_summary

expenses.head()
```

```python
transaction_summary = expenses.groupby('Transaction_Type')['Amount'].sum().reset_index()

print(transaction_summary)

total_revenue = transaction_summary.loc[transaction_summary['Transaction_Type'] == 'Revenue', 'Amount'].values[0]
total_expense = transaction_summary.loc[transaction_summary['Transaction_Type'] == 'Expense', 'Amount'].values[0]

profit_margin = ((total_revenue + total_expense) / total_revenue) * 100

print("Overall Profit Margin: {:.2f}%".format(profit_margin))

expenses.head()
```

```python
seasonal_pivot = expenses.groupby(['Season', 'Transaction_Type'])['Amount'].sum().unstack()

seasonal_pivot['Profit_Margin'] = ((seasonal_pivot['Revenue'] - seasonal_pivot['Expense']) / seasonal_pivot['Revenue']) * 100

print("Seasonal Profitability:")
print(seasonal_pivot)

expenses.head()
```

```python
expenses.columns = expenses.columns.str.strip()

print("First 5 rows of the dataset:")
print(expenses.head())

expenses = expenses.sort_values(by="Transaction_ID")

expenses["Transaction_Bin"] = expenses["Transaction_ID"] // 100 * 100

transaction_trends = expenses.groupby("Transaction_Bin")["Amount"].sum().reset_index()

print("Trends over Transaction ID bins:")
print(transaction_trends.head())

plt.figure(figsize=(10, 5))
plt.plot(transaction_trends["Transaction_Bin"], transaction_trends["Amount"], marker='o', linestyle='-')

plt.title('Expenditure/Revenue Trend Over Transaction Progression')
plt.xlabel('Transaction ID (Binned)')
plt.ylabel('Total Amount')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
if 'Incident_ID' in expenses.columns:

    incident_cost_summary = expenses.groupby('Incident_ID')['Amount'].mean().reset_index()
    print("Average Expenditure per Incident (first few records):")
    print(incident_cost_summary.head())
else:
    print("No Incident_ID column found in the dataset; skipping cost per incident analysis.")

```

```python
expenses.columns = expenses.columns.str.strip()

# Convert Amount to numeric (if not already)
expenses['Amount'] = pd.to_numeric(expenses['Amount'], errors='coerce')

# If the Season column does not exist and Date is available, derive Season.
if 'Season' not in expenses.columns and 'Date' in expenses.columns:
    expenses['Date'] = pd.to_datetime(expenses['Date'], errors='coerce')
    expenses['Season'] = expenses['Date'].dt.month.apply(
        lambda m: 'Winter' if m in [12, 1, 2] else ('Spring' if m in [3, 4, 5] else ('Summer' if m in [6, 7, 8] else 'Autumn'))
    )

print("Expenses dataset preview:")
print(expenses.head())
```

```python
import seaborn as sns

# Create a pivot table to get total Revenue and Expense per Season
seasonal_summary = expenses.groupby(['Season', 'Transaction_Type'])['Amount'].sum().unstack()
print("Seasonal Revenue vs. Expense:")
print(seasonal_summary)

# Calculate Profit Margin per Season: (Revenue + Expense) / Revenue * 100%
# Ensure that Revenue is positive
seasonal_summary['Profit_Margin'] = ((seasonal_summary['Revenue'] + seasonal_summary['Expense']) / seasonal_summary['Revenue']) * 100
print("Profit Margin by Season:")
print(seasonal_summary[['Profit_Margin']])

# Plot Profit Margin by Season
plt.figure(figsize=(8, 6))
sns.barplot(x=seasonal_summary.index, y=seasonal_summary['Profit_Margin'], palette="viridis")
plt.title("Profit Margin by Season")
plt.xlabel("Season")
plt.ylabel("Profit Margin (%)")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# Define fixed cost categories (update these as appropriate for your dataset)
fixed_categories = ['Staff Salary', 'Rent', 'Utilities']

# Create a new column "Cost_Type" based on whether the Category is fixed or variable
expenses['Cost_Type'] = expenses['Category'].apply(lambda x: 'Fixed' if x in fixed_categories else 'Variable')

# Group by Cost_Type and sum the Amounts
cost_type_summary = expenses.groupby('Cost_Type')['Amount'].sum()
print("Fixed vs Variable Cost Summary (raw values):")
print(cost_type_summary)

# Since expense amounts are negative, convert them to absolute values for pie chart plotting
cost_type_summary_abs = cost_type_summary.abs()
print("\nFixed vs Variable Cost Summary (absolute values):")
print(cost_type_summary_abs)

# Plot a pie chart of Fixed vs Variable Costs using the absolute values
plt.figure(figsize=(6, 6))
cost_type_summary_abs.plot(kind='pie', autopct='%1.1f%%', startangle=90, cmap='Pastel1')
plt.title("Fixed vs Variable Costs")
plt.ylabel("")  # Hide the y-label
plt.show()
```

```python
# Compute total fixed costs (take absolute value because expenses are negative)
fixed_cost = abs(expenses.loc[expenses['Cost_Type'] == 'Fixed', 'Amount'].sum())
# Compute total variable costs (absolute value)
variable_cost = abs(expenses.loc[expenses['Cost_Type'] == 'Variable', 'Amount'].sum())
# Compute total revenue (assumed to be positive)
total_revenue = expenses.loc[expenses['Transaction_Type'] == 'Revenue', 'Amount'].sum()

print("Fixed Costs (absolute):", fixed_cost)
print("Variable Costs (absolute):", variable_cost)
print("Total Revenue:", total_revenue)

if total_revenue > 0:
    CMR = (total_revenue - variable_cost) / total_revenue
    break_even_revenue = fixed_cost / CMR
    print(f"Contribution Margin Ratio (CMR): {CMR:.2f}")
    print(f"Break-even Revenue: {break_even_revenue:.2f}")
else:
    print("Total revenue is zero or not defined correctly; cannot compute break-even revenue.")
```

```python
# Filter for Equipment Maintenance costs
maintenance = expenses[expenses['Category'] == 'Equipment Maintenance']

# Check if Date is available and shows variation
if 'Date' in expenses.columns and expenses['Date'].nunique() > 1:
    # Convert Date to datetime if not already done
    maintenance['Date'] = pd.to_datetime(maintenance['Date'], errors='coerce')
    maintenance = maintenance.sort_values(by='Date')
    # Group by Year-Month
    maintenance_trend = maintenance.groupby(maintenance['Date'].dt.to_period('M'))['Amount'].sum().reset_index()
    maintenance_trend['Date'] = maintenance_trend['Date'].astype(str)
    
    plt.figure(figsize=(10, 5))
    plt.plot(maintenance_trend['Date'], maintenance_trend['Amount'], marker='o')
    plt.title('Monthly Trend of Equipment Maintenance Costs')
    plt.xlabel('Month')
    plt.ylabel('Total Maintenance Cost')
    plt.xticks(rotation=45)
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.show()
else:
    # If Date is not variable, use Transaction_ID bins as a proxy for time
    maintenance = maintenance.sort_values(by='Transaction_ID')
    maintenance["Transaction_Bin"] = maintenance["Transaction_ID"] // 100 * 100
    maintenance_trend = maintenance.groupby("Transaction_Bin")["Amount"].sum().reset_index()
    
    plt.figure(figsize=(10, 5))
    plt.plot(maintenance_trend["Transaction_Bin"], maintenance_trend["Amount"], marker='o')
    plt.title('Trend of Equipment Maintenance Costs Over Transactions')
    plt.xlabel('Transaction ID (Binned)')
    plt.ylabel('Total Maintenance Cost')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.show()
```

```python
# Filter only revenue transactions
revenue_data = expenses[expenses['Transaction_Type'] == 'Revenue'].copy()

# If 'Season' doesn't exist but 'Date' does, create Season from month
if 'Season' not in revenue_data.columns and 'Date' in revenue_data.columns:
    revenue_data['Date'] = pd.to_datetime(revenue_data['Date'])
    revenue_data['Season'] = revenue_data['Date'].dt.month.apply(
        lambda m: 'Winter' if m in [12,1,2] else
                  'Spring' if m in [3,4,5] else
                  'Summer' if m in [6,7,8] else 'Autumn'
    )

# Group by Season and Activity_Type to sum revenue
seasonal_revenue = revenue_data.groupby(['Season', 'Category'])['Amount'].sum().reset_index()
print("Seasonal Revenue by Ticket sales and Rentals:")
print(seasonal_revenue.head())
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='Season', y='Amount', hue='Category', data=seasonal_revenue)
plt.title("Seasonal Revenue by Ticket sales and Rentals")
plt.xlabel("Season")
plt.ylabel("Total Revenue")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
off_season_data = revenue_data[~revenue_data['Season'].isin(['Winter'])] 

off_season_summary = off_season_data.groupby('Category')['Amount'].sum().reset_index()
print("Off-Season Revenue by Activity:")
print(off_season_summary)
```

```python
plt.figure(figsize=(6, 6))
plt.pie(off_season_summary['Amount'], labels=off_season_summary['Category'], autopct='%1.1f%%', startangle=140)
plt.title("Off-Season Revenue Breakdown")
plt.show()
```
